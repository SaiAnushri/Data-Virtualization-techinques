## Load the OnlineRetail Dataset

import pandas as pd
import sqlite3

## Load the OnlineRetail Dataset

```python
df_retail = pd.read_csv('/content/OnlineRetail.csv', encoding='ISO-8859-1')
display(df_retail.head())
```

## Data Overview and Initial Cleaning

```python
print(df_retail.info())
print(df_retail.shape)
print(df_retail.columns)

# Clean column names (remove leading/trailing spaces)
df_retail.columns = df_retail.columns.str.strip()

# Convert object columns to lowercase (if applicable, though 'InvoiceNo' and 'StockCode' might be mixed case)
for col in df_retail.select_dtypes(include='object').columns:
    df_retail[col] = df_retail[col].astype(str).str.lower()

display(df_retail.head())
```

## Handle Missing Values

```python
print(df_retail.isnull().sum())

# Drop rows with any missing values
df_retail = df_retail.dropna()

print(f"Shape after dropping nulls: {df_retail.shape}")
```

## Handle Duplicate Rows

```python
print(f"Number of duplicated rows: {df_retail.duplicated().sum()}")

# Drop duplicate rows
df_retail = df_retail.drop_duplicates()

print(f"Shape after dropping duplicates: {df_retail.shape}")
```

## Descriptive Statistics

```python
print("\nDescriptive statistics for numerical columns:")
display(df_retail.describe())

print("\nDescriptive statistics for object columns:")
display(df_retail.describe(include='object'))
```

## Save Cleaned Data

```python
# Save to CSV
df_retail.to_csv("cleaned_online_retail.csv", index=False)

# Save to JSON
df_retail.to_json("cleaned_online_retail.json", orient="records", indent=4)
print("Cleaned data saved to 'cleaned_online_retail.csv' and 'cleaned_online_retail.json'.")
```

## Load Data into SQLite Database and Perform Queries

```python
# Establish SQLite connection and load data
connection_retail = sqlite3.connect("online_retail.db")
df_retail.to_sql("retail_transactions", connection_retail, if_exists="replace", index=False)
print(f"Successfully loaded {df_retail.shape[0]} records into 'retail_transactions' table.")
```

```python
print("\nFirst 5 rows from the database:")
display(pd.read_sql("SELECT * FROM retail_transactions LIMIT 5;", connection_retail))
```

```python
print("\nTransactions with Quantity greater than 10:")
display(pd.read_sql("SELECT * FROM retail_transactions WHERE Quantity > 10 LIMIT 5;", connection_retail))
```

```python
print("\nTransactions ordered by UnitPrice (descending), top 5:")
display(pd.read_sql("SELECT * FROM retail_transactions ORDER BY UnitPrice DESC LIMIT 5;", connection_retail))
```

```python
print("\nTotal transactions per Country:")
display(pd.read_sql("SELECT Country, COUNT(*) AS total_transactions FROM retail_transactions GROUP BY Country ORDER BY total_transactions DESC LIMIT 5;", connection_retail))
```

```python
# The error "ProgrammingError: Cannot operate on a closed database." indicates that
# the database connection 'connection_retail' was closed by a previous cell
# (likely the cell containing `connection_retail.close()`) before this cell tried to use it.
# To ensure this query runs, we will re-establish the connection.

connection_retail = sqlite3.connect("online_retail.db")

print("\nTotal number of records in the database:")
display(pd.read_sql("SELECT COUNT(*) AS total_records FROM retail_transactions;", connection_retail))
```

```python
print("\nAverage Quantity of items:")
display(pd.read_sql("SELECT AVG(Quantity) AS average_quantity FROM retail_transactions;", connection_retail))
```

```python
print("\nMaximum UnitPrice:")
display(pd.read_sql("SELECT MAX(UnitPrice) AS max_unit_price FROM retail_transactions;", connection_retail))
```

```python
print("\nMinimum UnitPrice:")
display(pd.read_sql("SELECT MIN(UnitPrice) AS min_unit_price FROM retail_transactions;", connection_retail))
```

```python
print("\nTotal sum of Quantity:")
display(pd.read_sql("SELECT SUM(Quantity) AS total_quantity FROM retail_transactions;", connection_retail))
```

```python
print("\nCountries with more than 1000 transactions:")
display(pd.read_sql("SELECT Country, COUNT(*) AS total FROM retail_transactions GROUP BY Country HAVING COUNT(*) > 1000 ORDER BY total DESC;", connection_retail))
```

```python
print("\nFirst 10 records from the database:")
display(pd.read_sql("SELECT * FROM retail_transactions LIMIT 10;", connection_retail))
```

```python
# Close the database connection
connection_retail.close()
print("SQLite connection closed.")
```
